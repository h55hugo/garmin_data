```python
import sqlite3
import pandas as pd

# ---- CONNECT ----
conn = sqlite3.connect("../db/garmin.db")

# ---- SELECT ACTIVITY ----
activity_id = 22903629287

df = pd.read_sql("""
    SELECT *
    FROM activities
    WHERE activityId = ?
""", conn, params=(activity_id,))

# ---- SHOW RESULT ----
print(df)

conn.close()
```

```python
# ---- CONNECT ----
conn = sqlite3.connect("../db/garmin.db")
cursor = conn.cursor()

# ---- DELETE ACTIVITY ----
activity_id = 22903629287

cursor.execute("""
    DELETE FROM activities
    WHERE activityId = ?
""", (activity_id,))

conn.commit()

print(f"Deleted activity {activity_id}")

conn.close()
```
